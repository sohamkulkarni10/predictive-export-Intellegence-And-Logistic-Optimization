````python
def analyze_news(news):

    prompt = f"""
You are an expert Agricultural Trade Intelligence Analyst.

The news may be written in ANY language.

Your tasks are:

1. Translate the news into English.
2. Give a short summary.
3. Identify the commodity.
4. Identify the country.
5. Determine whether this news indicates future import demand.
6. Extract structured information for machine learning.

Return ONLY valid JSON.

{{
"title_english":"",
"summary":"",

"commodity":"",

"country":"",

"event_type":"",

"demand_signal":"",

"demand_reason":"",

"market_impact":"",

"sentiment":"",

"sentiment_score":0,

"import_flag":0,

"export_flag":0,

"shortage_flag":0,

"production_drop":0,

"production_rise":0,

"price_increase":0,

"price_decrease":0,

"government_policy":0,

"weather_event":0,

"trade_issue":0,

"food_security_issue":0,

"confidence":0
}}

Rules

Event Type should be one of

Import Increase
Import Restriction
Export Increase
Export Restriction
Production Drop
Production Rise
Weather Disaster
Government Policy
Price Rise
Price Fall
Food Shortage
Trade Agreement
Conflict
Disease
Other

Demand Signal should be

High
Medium
Low
Neutral

Market Impact

Positive for Indian Exporters
Negative for Indian Exporters
Neutral

Sentiment

Positive
Neutral
Negative

Sentiment Score between -1 and 1

Flags must be 0 or 1.

Confidence between 0 and 1.

News

{news}

Return ONLY JSON.
"""

    response = client.chat.completions.create(

        model=MODEL,

        temperature=0,

        messages=[
            {
                "role":"user",
                "content":prompt
            }
        ]

    )

    text = response.choices[0].message.content

    text = text.replace("```json","")
    text = text.replace("```","")

    return json.loads(text)
````

````python
rows = []

for _, row in tqdm(df.iterrows(), total=len(df)):

    try:

        info = analyze_news(row["title"])

    except Exception as e:

        print(e)

        continue

    rows.append({

        "date": row["date"],

        "commodity_original": row["commodity"],

        "country_original": row["country"],

        "title_original": row["title"],

        "title_english": info["title_english"],

        "summary": info["summary"],

        "commodity": info["commodity"],

        "country": info["country"],

        "event_type": info["event_type"],

        "demand_signal": info["demand_signal"],

        "demand_reason": info["demand_reason"],

        "market_impact": info["market_impact"],

        "sentiment": info["sentiment"],

        "sentiment_score": info["sentiment_score"],

        "import_flag": info["import_flag"],

        "export_flag": info["export_flag"],

        "shortage_flag": info["shortage_flag"],

        "production_drop": info["production_drop"],

        "production_rise": info["production_rise"],

        "price_increase": info["price_increase"],

        "price_decrease": info["price_decrease"],

        "government_policy": info["government_policy"],

        "weather_event": info["weather_event"],

        "trade_issue": info["trade_issue"],

        "food_security_issue": info["food_security_issue"],

        "confidence": info["confidence"]

    })

    time.sleep(1)
````

````python
news_ml_dataset = pd.DataFrame(rows)

news_ml_dataset.to_csv("news_ml_dataset6.csv", index=False)

print(news_ml_dataset.head())
````
