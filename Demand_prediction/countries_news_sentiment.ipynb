````python
from deep_translator import GoogleTranslator

translator = GoogleTranslator(
    source="auto",
    target="en"
)
````

````python
def translate(text):

    if pd.isna(text):
        return ""

    try:
        return translator.translate(str(text))

    except:
        return str(text)
````

````python
df["title_english"] = df["title"].apply(translate)
````

````python
import time
import pandas as pd
from tqdm import tqdm
````
